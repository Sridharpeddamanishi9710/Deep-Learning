```python
import tensorflow as tf
from tensorflow import keras
import matplotlib.pyplot as plt
%matplotlib inline
import numpy as np
```

```python
(X_train,y_train),(X_test,y_test) = keras.datasets.mnist.load_data()
```

```python
X_train = X_train / 255
X_test = X_test / 255
```

```python
X_train[0]
```

```python
X_train_flattend = X_train.reshape(len(X_train),28*28)
```

```python
X_train_flattend.shape
```

```python
X_test.shape
```

```python
X_test_flatted = X_test.reshape(len(X_test),28*28)
```

```python
X_test_flatted.shape
```

```python
y_test[0]
```

```python
plt.matshow(X_train[0])
```

```python
y_train[:5]
```

```python
plt.matshow(X_train[1])
```

```python
model = keras.Sequential([
    keras.layers.Dense(10, input_shape=(784,),activation='sigmoid')
])

model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics = ['accuracy'] 
)

model.fit(X_train_flattend,y_train,epochs=5)
```

```python
model.evaluate(X_test_flatted,y_test)
```

```python
plt.matshow(X_test[2])
```

```python
y_pred = model.predict(X_test_flatted)
```

```python
y_pred[2]
```

```python
np.argmax(y_pred[2])
```

```python
y_pred_labels = [ np.argmax(i) for i in y_pred ]
```

```python
cm = tf.math.confusion_matrix(labels=y_test,predictions=y_pred_labels)
```

```python
y_pred_labels[:5]
```

```python
y_test[:5]
```

```python
import seaborn as sns
plt.figure(figsize=(10,7))
sns.heatmap(cm,annot=True,fmt='d')
plt.xlabel('Predicted')
plt.ylabel('Truth')
```

```python
model = keras.Sequential([
    keras.layers.Dense(100,input_shape=(784,),activation='relu'),
    keras.layers.Dense(10,activation='sigmoid')
])

model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

model.fit(X_train_flattend,y_train,epochs=5)
```

```python
model.evaluate(X_test_flatted,y_test)
```

```python
model = keras.Sequential([
    keras.layers.Flatten(input_shape=(28,28)),
    keras.layers.Dense(100,activation='relu'),
    keras.layers.Dense(10,activation='sigmoid')
])

model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

model.fit(X_train,y_train,epochs=5)
```

```python
model.evaluate(X_test,y_test)
```

```python
model1 = keras.Sequential([
    keras.layers.Flatten(input_shape=(28,28)),
    keras.layers.Dense(100,activation='relu'),
    keras.layers.Dense(100,activation='relu'),
    keras.layers.Dense(10,activation='softmax')
])

model1.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

model1.fit(X_train,y_train,epochs=5)
```

```python
model1.evaluate(X_test,y_test)
```

```python
model = keras.Sequential([
    keras.layers.Flatten(input_shape=(28,28)),
    keras.layers.Dense(100,activation='relu'),
    keras.layers.Dense(10,activation='sigmoid')
])

tb_callbacks = tf.keras.callbacks.TensorBoard(log_dir='logs/',histogram_freq=1)
model.compile(
    optimizer='sgd',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

model.fit(X_train,y_train,epochs=5,callbacks=[tb_callbacks])
```
