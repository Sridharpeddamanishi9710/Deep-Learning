```python
import numpy as np

y_true = [0,0,1,1,0]
y_predicted = [0.9,0.24,1,0.9,0]

def mae(y_true,y_predicted):
    total_error = 0
    for yp,yt in zip(y_true,y_predicted):
        total_error += abs(yt - yp)
    return total_error

mae(y_true,y_predicted)
```

```python
def mae(y_true,y_predicted):
    total_error = 0
    for yp,yt in zip(y_true,y_predicted):
        total_error += abs(yt - yp)
    print("Total Error:",total_error)
    mae =  total_error / len(y_true)
    print("MAE:",mae)
    return mae
```

```python
mae(y_true,y_predicted)
```

```python
def mse(y_true,y_predicted):
    total_squared_error = 0
    for yt,yp in zip(y_true,y_predicted):
        total_squared_error += (yt - yp) ** 2 
    print("total_squared_error",total_squared_error)
    mse = total_squared_error / len(y_true)
    return mse
```

```python
mse(y_true,y_predicted)
```

```python
def logloss(y_true,y_predicted):
    y_predicted_new = [max(i,epsilon) for i in y_predicted]
    y_predicted_new = [min(i,1-epsilon) for i in y_predicted_new]
    y_predicted_new = np.array(y_predicted_new)
    loss = -np.mean(y_true * np.log(y_predicted_new) + (1 - y_true) * np.log(1-y_predicted_new))
    return loss
```
