```python
model = gensim.models.Word2Vec(
    window=2,
    min_count=1,
    workers=4
)
```

```python
model.build_vocab(reviewText,progress_per=1000)
```

```python
model.train(reviewText, total_examples=model.corpus_count, epochs=model.epochs)
```
